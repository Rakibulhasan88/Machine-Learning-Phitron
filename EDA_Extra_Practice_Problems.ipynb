### Problem 3: Find Missing Values

```python
# Write your code here
# Count the missing values in each column.
df.isnull().sum()
```

```python
# Display only the columns that contain missing values.
df.columns[df.isnull().any()].tolist()
```

### Problem 5: Analyze Passenger Survival

1. Count how many passengers survived and how many did not survive.
2. Calculate the percentage for each group.
3. Create a Seaborn count plot for the `Survived` column.

```python
# Count how many passengers survived and how many did not survive.
df['Survived'].value_counts()
```

```python
# Calculate the percentage for each group.
(df['Survived'].value_counts()/len(df))* 100
```

```python
# Write your code here
# Create a Seaborn count plot for the `Survived` column
sns.countplot(data = df,x=df['Survived'])
plt.title('Counter plot of Survived people')
plt.xlabel('Count of Survived people')
plt.ylabel('Frequency')
plt.show()
```

### Problem 6: Analyze Categorical Features

For each of these columns — `Pclass`, `Sex`, and `Embarked`:

1. Count the number of records in each category.
2. Calculate the percentage of each category.
3. Create a separate Seaborn count plot.


```python
# Write your code here
# Count the number of records in each category.
cols_to_check = ['Pclass', 'Sex', 'Embarked']

for col in cols_to_check:
    print(df[col].value_counts(dropna=False))
    print()
```

```python
(df['Pclass'].value_counts()/len(df))* 100
```

```python
(df['Sex'].value_counts()/len(df))* 100
```

```python
(df['Embarked'].value_counts()/len(df))* 100
```

```python
# Create a Seaborn count plot for the `Pclass` column
sns.countplot(data = df,x=df['Pclass'])
plt.title('Counter plot Pclass')
plt.xlabel('Count of Pclass')
plt.ylabel('Frequency')
plt.show()
```

```python
# Create a Seaborn count plot for the `Sex` column
sns.countplot(data = df,x=df['Sex'])
plt.title('Counter plot Sex')
plt.xlabel('Count of Sex')
plt.ylabel('Frequency')
plt.show()
```

```python
# Create a Seaborn count plot for the `Embarked` column
sns.countplot(data = df,x=df['Embarked'])
plt.title('Counter plot Embarked')
plt.xlabel('Count of Embarked')
plt.ylabel('Frequency')
plt.show()
```
